# Import Libraries

```python
# load libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
# from sklearn import svm
# from sklearn.neighbors import KNeighborsClassifier
# from sklearn.ensemble import RandomForestClassifier
# from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

# Load Dataset

```python
loan_data = pd.read_csv("/Users/macbookpro/Desktop/ML & DL & GENAI Projects/Loan Approval Prediction/data/LoanApprovalPrediction.csv")
loan_data
```

# Data Preprocessing

```python
# first 5 rows
loan_data.head()
```

```python
# last 5 rows
loan_data.tail()
```

```python
# check for missing values
loan_data.isnull().sum()
```

```python
# number of columns of object datatypes
obj = (loan_data.dtypes == 'object')
print("Categorical Columns :", len(list(obj[obj].index)))
```

```python
# drop the Loan_ID column
loan_data.drop(['Loan_ID'], axis=1, inplace=True)
loan_data
```

```python
# convert the categorical columns into an int
label_encoder = preprocessing.LabelEncoder()
obj = (loan_data.dtypes == 'object')
for col in list(obj[obj].index):
    loan_data[col] = label_encoder.fit_transform(loan_data[col])
```

```python
# check number of columns with datatype == object
obj = (loan_data.dtypes == 'object')
print("Categorical Variables :", len(list(obj[obj].index)))
```

```python
# first 5 rows
loan_data.head()
```

## Married:       ## Education:             ## Self-Employed
## Yes ==> 1      ## Graduate ==> 0         ## Yes ==> 1
## No ==> 0       ## Not Graduate ==> 1     ## No ==> 0

```python
# check for missing values in columns
for col in loan_data.columns:
  loan_data[col] = loan_data[col].fillna(loan_data[col].mean()) 
 
loan_data.isna().sum()
```

```python
# convert the Loan_Status Column Y and N into 1 and 0
loan_data['Loan_Status'] = loan_data['Loan_Status'].replace({
    'Y': 1,
    'N': 0
})
```

```python
# first 5 rows
loan_data.head()
```

# Loan Status
## Yes ==> 1
## No ==> 0

```python
# correspondent values of Loan Status column
print(loan_data['Loan_Status'].value_counts())
```

# Data Splitting

```python
# split the features from the Target
X = loan_data.drop('Loan_Status', axis=1)
y = loan_data['Loan_Status']
```

```python
# data splitting
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# shape of the X and Y
X.shape, y.shape, X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

# Model Training

```python
model = LogisticRegression()
model.fit(X_train, y_train)
# classifier = svm.SVC(kernel='linear')
# classifier.fit(X_train, y_train)
```

# Model Evaluation

## Accuracy Score

```python
# accuracy score on training data
# X_train_pred = classifier.predict(X_train)
X_train_pred = model.predict(X_train)
train_data_acc = accuracy_score(X_train_pred, y_train)
```

```python
# result on the training data
print("Accuracy Score on Training Data: ", train_data_acc)
```

```python
# accuracy score on testing data
# X_test_pred = classifier.predict(X_test)
X_test_pred = model.predict(X_test)
test_data_acc = accuracy_score(X_test_pred, y_test)
```

```python
# result on the testing data
print("Accuracy Score on Testing Data: ",test_data_acc)
```

# Predictive System

```python
# input_data = (1,0,0.0,0,0,5849,0.0,144.968804,360.0,1.0,2) # APPROVED
input_data = (1,0,0.0,0,0,5849,0.0,144.968804,360.0,1.0,2) # NOT APPROVED

# change input data into a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the array for predicting 1 instance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

prediction = model.predict(input_data_reshaped)
# prediction = classifier.predict(input_data_reshaped)
print(prediction)

if (prediction[0] == 1):
    print("Loan NOT Approved")
else:
    print("Loan APPROVED")
```
